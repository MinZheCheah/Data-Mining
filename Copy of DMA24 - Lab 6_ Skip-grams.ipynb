### **Q5 (3 points)**
Using the vectors for the words in the Google News dataset, apply K-means clustering (K=2) and find the top 5 most representative words/phrases of each cluster.

*Note: Since there are ~3Mil words in the vocabulary, you can downsample it to 25k randomly selected words*  \\
*Hint: The "similar_by_vector" method might be useful*

**Do not delete the below cell**

```python
# Replace 0 with the code / value; Do not delete this cell
# YOUR CODE HERE
from sklearn.cluster import KMeans
import numpy as np
import random

random_words = random.sample(list(model.index_to_key), 25000)
word_vectors = np.array([model[word] for word in random_words])

kmeans = KMeans(n_clusters=2, random_state=42)
kmeans.fit(word_vectors)
cluster_centers = kmeans.cluster_centers_

most_rep_cluster1 = model.similar_by_vector(cluster_centers[0], topn=5)
most_rep_cluster2 = model.similar_by_vector(cluster_centers[1], topn=5)

most_rep_cluster1 = [word[0] for word in most_rep_cluster1]  # Extract only words, not similarity scores
most_rep_cluster2 = [word[0] for word in most_rep_cluster2]
```

```python
#This is an autograded cell, do not edit/delete
print(most_rep_cluster1)
print(most_rep_cluster2)
```

### **Q6 (1 point)**
What loss function does the skipgram model use and briefly describe what this function is minimizing.

**Do not delete the below cell**

The Skip-gram model utilizes the negative sampling loss function that is targeted toward the goal of learning word embeddings in a very efficient manner by distinguishing between real context words-those which in general appear within proximity to the target word in text, and "negative" samples, randomly chosen not to do so. It works in such a way that it maximizes the probability of the target word being similar to its actual context words while minimizing its probability of being similar to negative samples. The computation becomes less complex because, instead of all the words, it concentrates on a few negative samples. Negative sampling allows fast training of the model, focusing on a few negative samples instead of all possible words, and still manages to capture meaningful relationships between words.
